# 03 — Data Cleaning


```python
from pathlib import Path
import sys, pandas as pd
ROOT=Path.cwd().parent
if str(ROOT) not in sys.path: sys.path.insert(0,str(ROOT))
DATA=ROOT/'data'/'uploads'/'sample_business_data.csv'
df=pd.read_csv(DATA)
df.head()

from src.analysis.core import clean
clean_df=clean(df,drop_duplicates=True,clip_outliers=False)
clean_df.to_parquet(ROOT/'artifacts'/'cleaned_data.parquet',index=False)
clean_df.head()
```

```python
print("Original shape:", df.shape)
print("Cleaned shape:", clean_df.shape)
print("Missing values:", clean_df.isna().sum().sum())
print("Duplicate rows:", clean_df.duplicated().sum())
print("Saved file exists:", (ROOT / "artifacts" / "cleaned_data.parquet").exists())

clean_df.head()
```
